**Services:**

````python
// auth.service.ts
interface CurrentUser {
  userId: string;
  userName: string;
  email: string;
  roles: string[];
}

class AuthService {
  currentUser = signal<CurrentUser | null>(null);
  isAuthenticated = computed(() => !!this.currentUser());
  isMotard = computed(() => this.currentUser()?.roles.includes('Motard'));
  isParticipant = computed(() => this.currentUser()?.roles.includes('Participant'));
  
  login(email: string, password: string): Observable<void>
  logout(): Observable<void>
  checkAuth(): Observable<CurrentUser>  // GET /api/auth/me on app start
}
````
