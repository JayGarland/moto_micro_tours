Show the motard's riding history to build trust and credibility, while keeping it **clearly secondary** to upcoming tours.

---

### **Visual Hierarchy**

**Position:** After "Upcoming Tours" section, before "Reviews"

**Section Title:**
- "Completed Tours" or "Previous Rides"
- Font size: 24px (smaller than "Upcoming Tours")
- Color: `#424242` (medium gray, less prominent)

**Subtitle:**
- "Real rides I've done with other riders."
- Or "A glimpse of past routes and trips."
- Font size: 14px
- Color: `#757575` (light gray)
- Margin bottom: 16px

---

### **Card Design - Smaller & Focused**

**Card Size:**
- **Height:** ~120px (vs. 200px+ for upcoming tours)
- **Width:** 280px on desktop (horizontal scroll)
- **Width:** Full width on mobile (vertical stack)

**Card Style:**
- Background: `#fafafa` (lighter than white)
- Border: `1px solid #e0e0e0`
- Border radius: `8px` (softer than upcoming tours)
- Padding: `12px`
- **No hover elevation** (keeps it subtle)

**Card Content (Minimal):**

1. **Route/Area Name**
   - Font size: 16px
   - Font weight: 600
   - Color: `#212121`
   - Example: "Alsace Wine Route"

2. **Date or Month**
   - Font size: 13px
   - Color: `#757575`
   - Format: "Oct 2025" or "October 12, 2025"

3. **Distance / Duration**
   - Font size: 13px
   - Color: `#616161`
   - Icon + text: `🏁 180 km • 4h 30m`

4. **Highlight Badge (Optional)**
   - Small pill badge
   - Background: `#e3f2fd` (light blue)
   - Text color: `#0D47A1` (Yamaha blue)
   - Font size: 12px
   - Examples:
     - "Sunset ride"
     - "Beginner-friendly"
     - "Group of 6"
     - "Scenic route"

5. **CTA (Optional, Small)**
   - Text link (no button): "View details" or "See similar ride"
   - Font size: 13px
   - Color: `#0D47A1`
   - Underline on hover

## 🎯 **Smooth Responsive Layout - Best Practices**

### **The "Friction" Problem**

When resizing between desktop ↔ mobile, users often experience:
- ❌ Janky animations
- ❌ Elements jumping around
- ❌ Sticky sidebars misbehaving
- ❌ Layout "popping" between states

---

### **Solution: Single Breakpoint Strategy**

**Use ONE main breakpoint for major layout shift:**

```scss
$breakpoint-mobile: 768px;

// Mobile: ≤ 768px → Single column
// Desktop: > 768px → Two columns with sidebar
```

**This creates only TWO states:**

1. **State A (Mobile):** Full-width stack
2. **State B (Desktop):** Sidebar + main column

**Between these widths = just "zooming", not re-layout.**

---

### **Layout Implementation**

#### **Semantic HTML Order (for mobile-first):**

```html
<div class="profile-container">
  <!-- 1. Hero (always first visually on mobile) -->
  <app-profile-hero />
  
  <!-- 2. Sidebar (becomes top header on mobile) -->
  <app-profile-sidebar />
  
  <!-- 3. Main content -->
  <div class="main-content">
    <app-key-info-strip />
    <app-upcoming-tours />
    <app-completed-tours />
    <app-review-list />

  </div>
</div>
```

### **Avoid Layout "Friction"**

#### **❌ Don't Do This:**

```scss
// BAD: Animating layout properties
.profile-container {
  transition: all 0.3s ease; // ← CAUSES JANK
}

// BAD: Too many breakpoints
@media (max-width: 480px) { ... }
@media (max-width: 640px) { ... }
@media (max-width: 768px) { ... }
@media (max-width: 1024px) { ... }
@media (max-width: 1200px) { ... } // ← TOO MANY!
```

#### **✅ Do This:**

```scss
// GOOD: Only animate colors/opacity, not layout
.card {
  transition: background-color 0.2s ease, box-shadow 0.2s ease;
  // NO transition on width, flex-basis, grid-template-columns
}

// GOOD: One main breakpoint
@media (max-width: 768px) {
  // Mobile styles
}

@media (min-width: 769px) {
  // Desktop styles
}
```

---

### **Sticky Sidebar Best Practices**

#### **Mobile (≤ 768px):**
```scss
app-profile-sidebar {
  position: static; // No sticky behavior
  width: 100%;
  margin-bottom: 16px;
}
```

**Why?**
- Sticky sidebars on mobile cause overlaps
- Scrolling feels "janky" on small screens
- Users expect natural vertical flow

---

#### **Desktop (> 768px):**
```scss
@media (min-width: 769px) {
  app-profile-sidebar {
    position: sticky;
    top: 80px; // Offset for fixed navbar
    align-self: start;
    max-height: calc(100vh - 96px); // Prevent overflow
    overflow-y: auto; // Scrollable if content is tall
  }
}
```

**Why it works:**
- Wide screen = plenty of horizontal space
- Sidebar stays visible while scrolling main content
- Feels natural on desktop
