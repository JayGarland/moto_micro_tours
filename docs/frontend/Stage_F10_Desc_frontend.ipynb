### **1. Routes Implemented**

- **`/dashboard`** → Role-aware container that automatically redirects to:
  - `/dashboard/participant` for Participants
  - `/dashboard/motard` for Motards
- **`/dashboard/participant`** → Participant dashboard (protected by authGuard)
- **`/dashboard/motard`** → Motard dashboard (protected by authGuard)

All routes are lazily loaded and integrated into `app.routes.ts`.

### **2. Participant Dashboard Features**

#### **Data Management:**
- Uses `BookingsService.getParticipantBookings()` to load all bookings
- Computed signals for filtering:
  - `pendingRequests` - Bookings with status = Pending
  - `upcomingBookings` - Future bookings with status = Approved/Paid
  - `pastBookings` - Past bookings with status = Paid
  - `nextBooking` - Nearest upcoming booking

#### **UI Sections:**
1. **Dashboard Header**
   - Title: "My rides"
   - Subtitle: "Overview of your moto micro tours"
   - Primary action: "Browse tours" (navigates to `/tours`)

2. **Next Ride Card** (Featured/Highlighted)
   - Large prominent card with Yamaha blue gradient
   - Shows: tour title, date/time (French locale), motard name, price, status
   - Actions: "View booking details" button
   - Only displays if there's an upcoming booking

3. **Pending Requests Section**
   - Lists all pending booking requests awaiting motard approval
   - Empty state: "No pending requests" with ⏳ icon

4. **Upcoming Rides Section**
   - Lists all confirmed upcoming bookings (Approved/Paid status)
   - Empty state: "No upcoming rides" with 🏍️ icon

5. **Past Rides Section**
   - Lists completed rides
   - Empty state: "No past rides yet" with 📜 icon

### **3. Motard Dashboard Features**

#### **Data Management:**
- Loads data in parallel using `forkJoin`:
  - `BookingsService.getMotardBookingRequests()` - All bookings for motard
  - `ToursService.getTours()` - All tours (filtered for current user)
- Computed signals for filtering:
  - `pendingApprovals` - Bookings with status = Pending (awaiting decision)
  - `upcomingRides` - Future bookings with status = Approved/Paid
  - `myTours` - Tours where current user is a motard

#### **UI Sections:**
1. **Dashboard Header**
   - Title: "Motard dashboard"
   - Subtitle: "Manage your tours and bookings"
   - Primary action: "Create new tour" (navigates to `/tours/create`)

2. **Pending Booking Requests Section**
   - Special cards with blue border and top accent
   - Shows: participant request, tour details, request date, expiry
   - Status badge: "Waiting for your decision"
   - Action: "View booking" (navigates to booking detail)
   - Empty state: "No booking requests yet" with ✅ icon
   - Note: Approve/Decline buttons will be added in Stage F7

3. **Upcoming Rides Section**
   - Lists confirmed bookings where motard is the tour leader
   - Shows participant count and tour info
   - Empty state: "No upcoming rides" with 🏍️ icon

4. **My Tours Section**
   - Grid of tour cards showing:
     - Tour title and status (Active/Full/Completed/Draft)
     - Start → End locations
     - Date and time
     - Seats booked / Total capacity
     - Price range (min-max)
   - Action: "View tour" button
   - Empty state: "You haven't created any tours yet" with 🗺️ icon and "Create tour" action button
