
### **Stage F9 — Tour Creation (Motard Only)**
**Goal:** Motards can create tours

**Pages:**
1. **Create Tour Page** (`/tours/create`)
   - Form with:
     - Title, start date/time
     - Start/end locations
     - Notes/description
     - Initial motard (creator) capacity and price
   - "Publish" button

**Backend Endpoints Used:**
- `POST /api/tours`

**DoD:**
- Form validates (date in future, etc.)
- Only motards can access
- Tour appears in listing after creation
